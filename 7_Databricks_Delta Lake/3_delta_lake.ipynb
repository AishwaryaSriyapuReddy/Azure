# **Delta Lake Optimizations**

```python
df=spark.read.table("deltalakecatalog1.default.clonetbl")
```

```python
display(df)
```

```python
df.write.format("delta")\
    .mode("append")\
    .save("/Volumes/deltalakecatalog1/default/deltavol/optimization/")
```

```python
%sql
OPTIMIZE delta.`/Volumes/deltalakecatalog1/default/deltavol/optimization/`
```

```python
%sql
describe history delta.`/Volumes/deltalakecatalog1/default/deltavol/optimization/`
```

```python
%sql
select * from delta.`/Volumes/deltalakecatalog1/default/deltavol/optimization/`
where cust_id=1
```

```python
%sql
OPTIMIZE delta.`/Volumes/deltalakecatalog1/default/deltavol/optimization/` ZORDER BY cust_id;
```

```python
%sql
select * from delta.`/Volumes/deltalakecatalog1/default/deltavol/optimization/`
where cust_id=1
```

```python
%sql
describe history delta.`/Volumes/deltalakecatalog1/default/deltavol/optimization/`
```
